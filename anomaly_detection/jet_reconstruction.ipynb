```python
import uproot
import awkward as ak
import numpy as np
import matplotlib.pyplot as plt
```

```python
# filepath = "../data/SVJ_250.root"
# filepath = "../data/MinBias_PU200.root"
filepath = "../data/QCD_Pt15To3000_PU200.root"
# filepath = "../data/TT_PU200.root"
with uproot.open(filepath) as f:
    tree = f["outnano/Jets"]
    # Get the jet pt and eta arrays
    jet_hflav = tree["jet_genmatch_hflav"].array()
    jet_pflav = tree["jet_genmatch_pflav"].array()
    # jet_pt = tree["jet_pt"].array()
    # jet_eta = tree["jet_eta"].array()
    # jet_phi = tree["jet_phi"].array()

for flav in np.unique(jet_hflav):
    print(f"HF Flavor {flav:<2}: {np.sum(jet_hflav == flav):>7} jets")

for flav in np.unique(jet_pflav):
    print(f"HF Flavor {flav:<2}: {np.sum(jet_pflav == flav):>7} jets")

hf_options = [0, 4, 5]
pf_options = np.unique(np.abs(jet_pflav))

fig, ax = plt.subplots(1, 2, figsize=(8, 4))
# ax[0].hist(jet_hflav, bins=np.arange(-0.5, 6.5, 1), rwidth=0.8)
ax[0].bar(np.arange(len(hf_options)), [np.sum(jet_hflav == flav) for flav in hf_options], width=0.8)
ax[0].set_xticks(np.arange(len(hf_options)), labels=[str(flav) for flav in hf_options])
ax[0].set_xlabel('Jet HF Flavor')
ax[0].set_ylabel('Number of Jets')
# ax[1].hist(np.abs(jet_pflav), bins=np.arange(-0.5, 6.5, 1), rwidth=0.8)
ax[1].bar(np.arange(len(pf_options)), [np.sum(np.abs(jet_pflav) == flav) for flav in pf_options], width=0.8)
ax[1].set_xticks(np.arange(len(pf_options)), labels=[str(flav) for flav in pf_options])
ax[1].set_xlabel('Jet PF Flavor')
ax[1].set_ylabel('Number of Jets')
plt.tight_layout()
plt.show()

```

```python
# filepath = "../data/SVJ_250.root"
# filepath = "../data/MinBias_PU200.root"
# filepath = "../data/QCD_Pt15To3000_PU200.root"
filepath = "../data/TT_PU200.root"
with uproot.open(filepath) as f:
    tree = f["outnano/Jets"]
    # Get the jet pt and eta arrays
    # jet_hflav = tree["jet_genmatch_hflav"].array()
    # jet_pflav = tree["jet_genmatch_pflav"].array()
    # jet_pt = tree["jet_pt"].array()
    pt_arr = tree["jet_puppicand_pt"].array()
    ptlog_arr = tree["jet_puppicand_pt_log"].array()
    eta_arr = tree["jet_puppicand_deta"].array()
    phi_arr = tree["jet_puppicand_dphi"].array()

eta = ak.flatten(eta_arr)
phi = ak.flatten(phi_arr)

plt.hist(eta, bins=100, range=(-200, 200), histtype='step', label='Eta')
plt.hist(phi, bins=100, range=(-200, 200), histtype='step', label='Phi')
plt.yscale('log')
plt.xlabel('Radians')
plt.ylabel('Number of Jets')
plt.legend()
plt.show()
```

```python
eta_phys = eta * (np.pi / 720.0)
phi_phys = phi * (np.pi / 720.0)

theta = np.random.uniform(low=0.0, high=2 * np.pi, size=np.shape(eta)[0])
cos_theta = np.cos(theta)
sin_theta = np.sin(theta)

eta_phys_rot = cos_theta * eta_phys - sin_theta * phi_phys
phi_phys_rot = sin_theta * eta_phys + cos_theta * phi_phys

eta_rot = np.round(eta_phys_rot * (720.0 / np.pi))
phi_rot = np.round(phi_phys_rot * (720.0 / np.pi))

fig, ax = plt.subplots(2, 1, figsize=(8, 6), gridspec_kw={'height_ratios': [3, 1]})

ax[0].hist(eta, bins=100, range=(-100, 100), histtype='step', label='Eta', color='b')
ax[0].hist(phi, bins=100, range=(-100, 100), histtype='step', label='Phi', color='r')
ax[0].hist(eta_rot, bins=100, range=(-100, 100), histtype='step', label='Eta Rotated', color='tab:blue')
ax[0].hist(phi_rot, bins=100, range=(-100, 100), histtype='step', label='Phi Rotated', color='tab:red')
ax[0].set_ylim(1e3, 1e6)
ax[0].set_yscale('log')
ax[0].set_xlabel('Discrete Units')
ax[0].set_ylabel('Number of Jets')
ax[0].legend()

cn, bn = np.histogram(eta, bins=100, range=(-100, 100))
cd, bd = np.histogram(eta_rot, bins=100, range=(-100, 100))
ax[1].scatter(bn[:-1], cd / cn, alpha=0.5, label='Eta', color='b', s=2)

cn, bn = np.histogram(phi, bins=100, range=(-100, 100))
cd, bd = np.histogram(phi_rot, bins=100, range=(-100, 100))
ax[1].scatter(bn[:-1], cd / cn, alpha=0.5, label='Phi', color='r', s=2)

ax[1].set_xlabel('Discrete Units')
ax[1].set_ylabel('Number of Jets')
ax[1].grid()

ax[1].legend()

```

```python
pt = ak.flatten(pt_arr)

pt = np.logspace(np.log(1), np.log(1000), num=1000)
sigma = np.clip(0.01 + 0.05 / np.sqrt(pt), 0.015, 0.075)
pt_aug = pt * np.exp(np.random.normal(loc=0.0, scale=sigma, size=pt.shape))

plt.plot(pt, sigma, label='Sigma', color='b')
plt.scatter(pt, pt_aug / pt - 1, alpha=0.5, s=10, label='Augmented Pt', color='r')
plt.xscale('log')
plt.xlabel('Pt')
plt.ylabel('Relative Pt Difference')    
```
